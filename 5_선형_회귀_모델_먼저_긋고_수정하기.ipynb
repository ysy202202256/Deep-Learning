```python
#단순 선형 회귀(변수가 한 개)

import numpy as np
import matplotlib.pyplot as plt

#공부 시간 x와 성적 y의 넘파이 배열
x = np.array([2,4,6,8])
y = np.array([81,93,91,97])

#데이터의 분포를 그래프로 나타내기
#plt.scatter(x,y)
#plt.show()

#a, b의 값 초기화
a=0
b=0

#학습률
Ir = 0.03

#반복 횟수
epochs = 2001

#x값의 수
n = len(x)

#Gradient Descent(경사하강법) 시작
for i in range(epochs):
  y_pred = a * x + b #예측값 계산
  error = y - y_pred #오차 계산

  a_diff = (2/n) * sum(-x * (error)) #오차함수를 a로 편미분
  b_diff = (2/n) * sum(-(error)) #오차함수를 b로 편미분

  a = a - Ir * a_diff #a값 업데이트
  b = b - Ir * b_diff #b값 업데이트

  if i % 100 == 0:
    print("epoch = %.f, 기울기=%.04f, 절편=%.04f" %(i,a,b))

#최종 그래프
y_pred = a * x + b

#그래프 그리기
plt.scatter(x,y)
plt.plot(x, y_pred, 'r')
plt.show()
```

```python
#다중 선형 회귀(변수가 여러 개)

#공부한 시간 x1, 과외 수업 횟수 x2, 성적 y를 넘파이 배열로 생성
x1 = np.array([2,4,6,8])
x2 = np.array([0,4,2,3])
y = np.array([81,93,91,97])

a1 = 0
a2 = 0
b = 0
Ir = 0.01
epochs = 2001

y_pred = a1 * x1 + a2 * x2 + b #기울기와 절편 변수 설정
error = y - y_pred #오차 계산

n = len(x1)

for i in range(epochs):
  y_pred = a1 * x1 + a2 * x2 + b
  error = y - y_pred
  a1_diff = (2/n) * sum(-x1 * (error))
  a2_diff = (2/n) * sum(-x2 * (error))
  b_diff = (2/n) * sum(-(error))

  a1 = a1 - Ir * a1_diff
  a2 = a2 - Ir * a2_diff
  b = b - Ir * b_diff

  if i % 100 == 0:
    print("epoch = %.f, 기울기1=%.04f, 기울기2=%.04f, 절편=%.04f" %(i,a1,a2,b))

print("실제 점수:", y)
print("예측 점수:", y_pred)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

x = np.array([2, 4, 6, 8])
y = np.array([81, 93, 91, 97])

model = Sequential()
model.add(Dense(1, input_dim=1, activation='linear'))
model.compile(optimizer='sgd', loss='mse')
model.fit(x, y, epochs=2000)

plt.scatter(x, y)
plt.plot(x, model.predict(x), 'r')
plt.show()

hour = 7
input_data = tf.constant([[hour]])
prediction = model.predict(input_data)[0][0]
print("%.f시간을 공부할 경우의 예상 점수는 %.02f점입니다." % (hour, prediction))
```

```python
x = np.array([[2, 0], [4, 4], [6, 2], [8, 3]])
y = np.array([81, 93, 91, 97])

model = Sequential()

# 입력 변수가 2개(학습 시간, 과외 시간)이므로 input_dim에 2를 입력합니다.
model.add(Dense(1, input_dim=2, activation='linear'))
model.compile(optimizer='sgd' ,loss='mse')

model.fit(x, y, epochs=2000)

hour = 7
private_class = 4

input_data = tf.constant([[hour, private_class]])
prediction = model.predict(input_data)[0][0]
print("%.f시간을 공부하고 %.f시간의 과외를 받을 경우, 예상 점수는 %.02f점입니다" % (hour, private_class, prediction))
```
