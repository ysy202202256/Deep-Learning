```python
import numpy as np

#공부한 시간과 점수를 각각 x, y라는 이름의 넘파이 배열로 만든다.
x = np.array([2,4,6,8])
y = np.array([81,93,91,97])

#x의 평균값을 구한다.
mx = np.mean(x)

#y의 평균값을 구한다.
my = np.mean(y)

#각각의 평균을 출력
print("x의 평균값:", mx)
print("y의 평균값:", my)

#분모 구하기
divisor = sum([(i-mx)**2 for i in x])

#분자 구하기
def top(x, mx, y, my):
  d=0
  for i in range(len(x)):
    d += (x[i]-mx) * (y[i]-my)
  return d
dividend = top(x,mx,y,my)

#각각을 출력
print("분모:", divisor)
print("분자:", dividend)

#최소제곱직선의 기울기 a
a = dividend / divisor

#y절편 b 구하기
b = my - (mx*a)

print(f"따라서, 주어진 자료의 최소제곱직선은 y={a}x+{b}이다.")
```

```python
#가상의 기울기와 y절편 결정
fake_a = 3
fake_b = 76

#공부 시간 x와 성적 y의 넘파이 배열 생성
x = np.array([2,4,6,8])
y = np.array([81,93,91,97])

#y = ax + b에 가상의 a값과 b값을 대입한 결과를 출력하는 함수
def predict(x):
  return fake_a * x + fake_b

#예측 값이 들어갈 빈 리스트 생성
predict_result = []

#모든 x값을 한 번씩 대입하여 predict_result 리스트 완성
for i in range(len(x)):
  predict_result.append(predict(x[i]))
  print("공부시간=%.f, 실제점수=%.f, 예측점수=%.f" %(x[i], y[i], predict(x[i])))

#평균 제곱 오차 함수
n = len(x)
def mse(y, y_pred):
  return (1/n) * sum((y-y_pred)**2)

#평균 제곱 오차 출력
print("평균 제곱 오차: " + str(mse(y, predict_result)))
```
